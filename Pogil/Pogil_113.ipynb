```python
%%writefile main.cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> values = { 1, 3, 5, 6, 7, 9 };

    for (int &each : values) {
        // add 7 to each element
        each += 7;
    }
    for (auto each : values) {
        cout << each << ' ';
    }
    cout << endl;

    return 0;
}
```

## Model 3: `map`

In Python we learned about a `Dictionary` that maps unique keys to values. C++ has a similar container, called a `Map` with the following common methods:

| Method | Description |
| ------ | ----------- |
| emplace(key, value) | Associates key with specified value. If key already exists, the map entry is not changed. |
| at(key) | Returns the entry associated with key. If key does not exist, throws an out_of_range exception. |
| count(key) | Returns 1 if key exists, otherwise returns 0. |
| erase(key) | Removes the map entry for the specified key if the key exists. |
| clear() | Removes all map entries. |

Run the following code and determine what it does.

```python
%%writefile main.cpp
#include <iostream>
#include <map>
#include <utility> // For pair
using namespace std;

int main() {
    // Creating a map of int keys and string values
    map<int, string> myMap;

    // Inserting key-value pairs into the map
    myMap[1] = "One";
    myMap[2] = "Two";
    myMap[3] = "Thee";

    // Accessing and printing the values
    for (const auto& pair : myMap) {
        cout << "Key: " << pair.first << ", Value: " << pair.second << endl;
    }

    return 0;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```

## Model 4: `set`

Run the following code and observe the output.

```python
%%writefile main.cpp
#include <iostream>
#include <set>
#include <vector>
using namespace std;

int main() {
    // Initializing a vector with duplicate values
    vector<int> myVector = {1, 2, 3, 2, 4, 5, 3, 6};

    // Initializing a set with the same values
    set<int> mySet = {1, 2, 3, 2, 4, 5, 3, 6};

    // Printing the contents of the vector
    cout << "Vector contents: ";
    for (const int& value : myVector) {
        cout << value << " ";
    }
    cout << endl;

    // Printing the contents of the set
    cout << "Set contents: ";
    for (const int& value : mySet) {
        cout << value << " ";
    }
    cout << endl;

    return 0;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```
