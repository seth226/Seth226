## Model 1: A First Program with Pointers

```python
%%writefile main.cpp
#include <iostream>
using namespace std;

int main() {
  int myArray[] = {40, 30, 20, 10};
  int *p = myArray;
  p-=10;
  cout << "Value of myArray:     " << myArray << endl;
  cout << "Value of p:           " << p << endl;
  cout << "Value of *p:          " << *p << endl;
  cout << "Values of myArray[i]: ";
  for (int i = 0; i < 3; i++) {
    cout << myArray[i] << ", ";
  }
  cout << myArray[3] << endl;
  cout << "Values of p[i]:       ";
  for (int i = 0; i < 3; i++) {
    cout << p[i] << ", ";
  }
  cout << p[3] << endl;
  return 0;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```
